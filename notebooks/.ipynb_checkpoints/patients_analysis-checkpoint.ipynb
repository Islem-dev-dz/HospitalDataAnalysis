```python
# 1. Loading the data :

# Import the pandas library, which is used to easily manipulate data tables (DataFrames)
import pandas as pd

# Read the "patients.csv" file located in the "data" folder and load its content into a DataFrame named 'df'
df = pd.read_csv("../data/patients.csv")

# Display the first few rows of the table to visually check the content and structure of the data
df.head()
```

```python
# 2. Descriptive statistics :

# Generate a global statistical summary of the numerical columns (mean, min, max, standard deviation, quartiles, ...)
df.describe()

# Count the number of occurrences for each gender (how many men and women are in the table)
df['Sex'].value_counts()

# Count the number of patients for each medical diagnosis type (Normal, Hypertension, Diabetes)
df['Diagnosis'].value_counts()
```

```python
# 3. Data visualization :

# Import the plotting module from the matplotlib library
import matplotlib.pyplot as plt

# Create a line plot connecting the data points : Age on the X-axis and Cholesterol on the Y-axis, with circular markers on each point
plt.plot(df['Age'], df['Cholesterol'], marker = "o")

# Set the label displayed on the horizontal axis (X-axis)
plt.xlabel('Age')

# Set the label displayed on the vertical axis (Y-axis)
plt.ylabel('Cholesterol')

# Assign a title to the chart
plt.title('Cholesterol vs Age')

# Display the generated plot on the screen
plt.show()
```
